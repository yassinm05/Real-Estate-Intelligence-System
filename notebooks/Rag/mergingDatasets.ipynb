```python
import pandas as pd
import uuid
import os
```

```python
listingPath = r"C:\Users\w\Desktop\real_estate_rag_ecosystem\data\cleaned_airbnb_listings.csv"
reviewsPath = r"C:\Users\w\Desktop\real_estate_rag_ecosystem\data\cleaned_airbnb_reviews.csv"
listings_df = pd.read_csv(listingPath)
reviews_df = pd.read_csv(reviewsPath)
```

```python
listings_df['id'] = listings_df['id'].astype(str)
reviews_df['listing_id'] = reviews_df['listing_id'].astype(str)
```

```python
merged_df = pd.merge(
    reviews_df, 
    listings_df, 
    left_on='listing_id',  
    right_on='id',        
    how='inner'
)
```

```python
merged_df.shape
```

```python
merged_df['chroma_review_id'] = [str(uuid.uuid4()) for _ in range(len(merged_df))]
```

```python
merged_df.columns
```

```python
merged_df = merged_df.drop(columns=['id'])
```

```python
outputpath = r'C:\Users\w\Desktop\real_estate_rag_ecosystem\data\merged_airbnb_data.csv'

merged_df.to_csv(outputpath, index=False)
```
